```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, element):
        self.items.append(element)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def size(self):
        return len(self.items)

    def is_empty(self):
        return self.items == []


def reverse_string_with_stack(text):
    stack = Stack()

    for char in text:
        stack.push(char)

    reversed_text = ""
    while not stack.is_empty():
        reversed_text += stack.pop()

    return reversed_text


print(reverse_string_with_stack("Rafael"))
print(reverse_string_with_stack("stars"))
print(reverse_string_with_stack("cool"))
```

```python
# Custom Stack (from scratch)
class CustomStack:
    class Node:
        def __init__(self, data):
            self.data = data
            self.below = None

        def __str__(self):
            return f"Node ({self.data})"

    def __init__(self):
        self.top = None

    def is_empty(self):
        return self.top is None

    def push(self, value):
        if self.is_empty():
            self.top = self.Node(value)
        else:
            new_node = self.Node(value)
            new_node.below = self.top
            self.top = new_node

    # mini Challenge
    def pop(self):
        if self.is_empty():
            raise IndexError("attempting pop() on empty stack")

        held_node = self.top
        self.top = held_node.below
        return held_node.data

    # Assignment 3
    def peek(self):
        if self.is_empty():
            raise IndexError("peek() on empty stack")
        return self.top.data

    def size(self):
        count = 0
        current = self.top
        while current is not None:
            count += 1
            current = current.below
        return count

    def __str__(self):
        values = []
        current = self.top
        while current is not None:
            values.append(str(current.data))
            current = current.below
        return f"CustomStack - [{', '.join(values)}]"
```

# Queues
A queue is an ordered collection of elements where items are added and removed from  **opposite** ends, commonly referred to as the  "front" or "rear" of the queue. The ordering principle is referred to as FIFO (First In, First Out) or "first come, first served".

```python
# Simplified Queue (relying on built-ins)

class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, element):
        self.items.insert(0, element)

    def dequeue(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items)-1]

    def size(self):
        return len(self.items)

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Queue - ({self.items})"
```

```python
q1 = Queue()

q1.enqueue("Apple")
q1.enqueue("Watemerlon")
q1.enqueue("Lime")
q1.enqueue("Strawberry")

print(q1)
q1.peek()
q1.size()
q1.is_empty()
x =q1.dequeue()
print(x)
print(q1)
```

# Mini Challenge 3 - Reverse a list using a queue

"""
You're given a list = numbers = [1,2,3,4,5]
Create a function using a queue, reverse the orders of the numbers and return the reversed list
"""

```python
def reversing_list(numbers):
    q1 = Queue()

    for number in numbers:
        q1.enqueue(number)

    reversed_numbers = []
    while not q1.is_empty():
        reversed_numbers.insert(0, q1.dequeue())

    return reversed_numbers


numbers = [1, 2, 3, 4, 5]
print(reversing_list(numbers))
```
