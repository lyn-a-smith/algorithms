# OOP (Object Oriented Programming)
Python is a pbject-oriented language, allowing us to structutr code using classess and objects for better organization and reusability.

## Andvantages
- Provides a clear structure to programs
- Makes code easier to mantain, reuse and debug
- Helps keep your code DRY (Do not Repeat Yourself)
- Allows to rebuild reusable applications with less code

## Classes and Objects
A class defines what an object looks like and its behavior.

An object is created based on that class.

For example:

    Class -> Fruit
    Objects -> Apple, Banna, Mango
    Class -> Car
    Objects -> Audi, Tesla, Toyota

### The four pillars of OOP
1. Encapsulation
2. Abstraction
3. Inheritance
4. Polymorphism

````python
"""
Syntax
Keyword 'class' + name of the class
Note: The name of the class should follow Pascal Case notation e.d. MyPersona
Construcro - Is always execute when the class is being initialized
Propeties/Attributes - The variables that help us to describe our clsass
Method - Function that belong to the class 
"""

class Person:
    # Constructor - Executes when initializing an object
    def __init__(self, name, last_name, age):
        self.name = name
        self.last_name = last_name
        self.age = age

    def greeting(self):
        return f"Hello from {self.name}"

    def __str__(self):
        return f"Person(name={self.name}, last_name={self.last_name}, age={self.age})"
````

````python
person1 = Person("Lyn", "Smith", 25)
print(person1)
print(person1.name)

````

# Pseudocode Challenge 2: Find the largest number in a list

```
START
  INPUT numbers_list
  IF numbers_list is empty THEN
    PRINT "The list is empty. No largest number."
    STOP
  END IF

  SET largest_number = numbers_list[0]
  SET i = 1

  WHILE i < LENGTH(numbers_list) DO
    SET current_number = numbers_list[i]
    IF current_number > largest_number THEN
      SET largest_number = current_number
    END IF
    SET i = i + 1
  END WHILE

  PRINT "The largest number is " + largest_number
END
```

Plan:
- Start by receiving the list of numbers.
- If the list is empty, print a message and stop.
- SET `largest_number` with the first item in the list.
- Loop through each remaining numbers in the list.
- For each number, compare it to `largest_number`.
- If the current number is bigger, update `largest_number`.
- After the loop finishes, print the final `largest_number`.

````python
# Challenge 2 with collection
numbers = [3, 5, 27, 20, 6, 3, 5, 60]

if all(isinstance(number, (int, float)) for number in numbers):
    largest = numbers[0]

    for number in numbers:
        if number > largest:
            largest = number

    print("The largest number is:", largest)
else:
    print("The collection does not contain only numbers.")
````
