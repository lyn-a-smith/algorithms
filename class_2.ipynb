## Anagram Create a function that takes two strings as parameters and returns True if they are anagrams of each other, False otherwise.

    Examples:
    1. heart and earth
    2 scar and cars
    3. star and rats
    etc

## Criteria
Two strings are anagrams of earch other if and only if all the criteria below are met:

    1. They must ause all the same characters
    2. They must be exactly the same length (in characters)
    3. No sorted() allowed

```python
def is_anagram(str1, str2):
    if len(str1) != len(str2):
        return False

    counts = {}

    for char in str1:
        counts[char] = counts.get(char, 0) + 1

    for char in str2:
        if char not in counts:
            return False
        counts[char] -= 1
        if counts[char] < 0:
            return False

    return all(value == 0 for value in counts.values())


print(is_anagram("heart", "earth"))
print(is_anagram("scar", "cars"))
print(is_anagram("star", "rats"))
print(is_anagram("hello", "world"))
print(is_anagram("ratt","star"))
```

## Stacks 

A stack is an ordered collection of elements where items are added and removed from one end, commonly referred to as the "top". The ordering principle is referred to ad LIFO (last in, first out)

```python
# Simplified version (relying on built-in functions)

class Stack:
    def __init__(self):
        self.items = []

    def push(self, elements):
        self.items.append(elements)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items)-1]

    def size(self):
        return len(self.items)

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Stack - {self.items}"
```

```python
stack_obj = Stack()
stack_obj.is_empty()

stack_obj.push("A")
stack_obj.push("B")
stack_obj.push("C")
print(stack_obj)
top_value = stack_obj.pop()
print(top_value)
print(stack_obj)
print(stack_obj.peek())
```

```python
# Assignment 2

class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def multiply(self):
        return self.a * self.b

    def divide(self):
        if self.b == 0:
            return "Cannot divide by zero"
        return self.a / self.b

    def __str__(self):
        return f"Calculator(a={self.a}, b={self.b})"


c1 = Calculator(5, 2)
print(c1.add())
print(c1.multiply())
print(c1.subtract())
print(c1.divide())
print(c1)
```
